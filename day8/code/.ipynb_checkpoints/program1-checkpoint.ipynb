```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
```

```python
# load all the keys from .env file
load_dotenv()
```

```python
# create llm object
llm = ChatOpenAI(model="gpt-5", temperature=1.0)
```

```python
# send a prompt and get the response
result = llm.invoke("what is the current temperature in Pune")
result
```

```python
# create a tool to get current temperature of a city
@tool
def get_current_temperature(city: str):
    """Used to get the current temperature of a city"""
    return 20
```

```python
get_current_temperature.invoke({"city": "pune"})
```

```python
# create a tool to multiply two numbers
@tool
def multiply(p1: int, p2: int) -> int:
    """Used to multiply two numbers"""
    return p1 * p2
```

```python
@tool
def generate_city_data(city: str) -> str:
    """Used to get data generated about a city"""
    return "this is a data about the city"
```

```python
# list of tools
tools = [get_current_temperature, multiply, generate_city_data]

# bind the tool with llm so that llm can use this tool when needed
llm_with_tools = llm.bind_tools(tools=tools)
```

```python
# send a prompt and get the response
result = llm_with_tools.invoke("what is the current temperature in Pune")
result.tool_calls
```

```python
# send a prompt and get the response
result = llm_with_tools.invoke("what is the current temperature in New York")
result.tool_calls
```

```python
# send a prompt and get the response
result = llm_with_tools.invoke("what is 20 * 20")
result.tool_calls
```

```python
# send a prompt and get the response
result = llm_with_tools.invoke("generate data about pune")
result
```

```python
result = llm_with_tools.invoke("Pune temperature")
result.tool_calls
```

```python
result = llm_with_tools.invoke("water temperature")
result.tool_calls
```
